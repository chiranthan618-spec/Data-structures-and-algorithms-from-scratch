````python
#Object oriented code for Binary tree data Structure
from typing import Any #Importing this to type hint the variable to random data type
class Node:
    def __init__(self, data: Any): 
        self.data = data 
        self.right = None
        self.left = None
````

Types of binary trees:
1) Full Binary Tree:
    All the tree nodes have either 0 or 2 children.
 ```
   __50__ 
  /      \
 30      70
/  \       
20 40      MA
```
2) Complete Binary tree: 
All the levels are completely filled except for the last, which is left filled.
```mermaid
flowchart TD
    A((1)) --> B((2))
    A --> C((3))
    B --> D((4))
    B --> E((5))
    C --> F((6))

```
3) Perfect binary tree: All the levels are completely filled.


NOTE: 
A leaf node is the one with zero children nodes(Zero degree). Now for any binary tree:
$$Number\ of\ leaf\ node\ =\ number\ of\ nodes\ with\ degree\ 2\ +\ 1\$$

In a perfect binary tree, 
1) the ith level always has $2^i$ nodes.
2) Number of internal nodes are $2^{h-1}$
3) Total number of nodes = $2^{h+1}\ -1$



Different ways of traversing through a binary tree:
1) Pre Order traversal: use of recursion
```
   __50__ 
  /      \
 30      70
/  \       
20 40      
```
Here the out put will be: [50, 30, 20, 40, 70]

2) Post order traversal: Use of recursion
```
   __50__ 
  /      \
 30      70
/  \       
20 40  
```    
Output: [20, 40, 30, 70, 50]

3) In Order traversal: Use of recursion
```
   __50__ 
  /      \
 30      70
/  \       
20 40      
```
output: [20, 30, 40, 50, 70]

4) Level Order traversal: Use of queue to access this traversal.
```
   __50__ 
  /      \
 30      70
/  \       
20 40
```      
here, the output should return a list with each level stored in a list.
output: [[50], [30, 70], [20, 40]]

````python
class PreOrderTraversal:
    def __init__(self):
        self.ans = []
    def PreOrder(self, root):
        if root is None:
            return
        else:
            self.ans.append(root.data)
            self.PreOrder(root.left)
            self.PreOrder(root.right)
    def traversal(self,root):
        self.PreOrder(root)
        return self.ans
````

````python
class PostOrderTraversal:
    def __init__(self):
        self.ans = []
    def PostOrder(self, root):
        if root is None:
            return
        else:
            self.PostOrder(root.left)
            self.PostOrder(root.right)
            self.ans.append(root.data)
    def traversal(self,root):
        self.PostOrder(root)
        return self.ans
````

````python
class InOrderTraversal:
    def __init__(self):
        self.ans = []
    def InOrder(self, root):
        if root is None:
            return 
        else:
            self.InOrder(root.left)
            self.ans.append(root.data)
            self.InOrder(root.right)
    def traversal(self, root):
        self.InOrder(root)
        return self.ans
````
