```python
#Object oriented code for Binary tree data Structure
from typing import Any #Importing this to type hint the variable to random data type
class Node:
    def __init__(self, data: Any): 
        self.data = data 
        self.right = None
        self.left = None
```

Different ways of traversing through a binary tree:
1) Pre Order traversal: use of recursion
   __50__ 
  /      \
 30      70
/  \       
20 40      
Here the out put will be: [50, 30, 20, 40, 70]

2) Post order traversal: Use of recursion
   __50__ 
  /      \
 30      70
/  \       
20 40      
Output: [20, 40, 30, 70, 50]

3) In Order traversal: Use of recursion
   __50__ 
  /      \
 30      70
/  \       
20 40      
output: [20, 30, 40, 50, 70]

4) Level Order traversal: Use of queue to access this traversal.
   __50__ 
  /      \
 30      70
/  \       
20 40      
here, the output should return a list with each level stored in a list.
output: [[50], [30, 70], [20, 40]]

```python
class PreOrderTraversal:
    def __init__(self):
        self.ans = []
    def PreOrder(self, root):
        if root is None:
            return
        else:
            self.ans.append(root.data)
            self.PreOrder(root.left)
            self.PreOrder(root.right)
    def traversal(self,root):
        self.PreOrder(root)
        return self.ans
```

```python
class PostOrderTraversal:
    def __init__(self):
        self.ans = []
    def PostOrder(self, root):
        if root is None:
            return
        else:
            self.PostOrder(root.left)
            self.PostOrder(root.right)
            self.ans.append(root.data)
    def traversal(self,root):
        self.PostOrder(root)
        return self.ans
```

```python
class InOrderTraversal:
    def __init__(self):
        self.ans = []
    def InOrder(self, root):
        if root is None:
            return 
        else:
            self.InOrder(root.left)
            self.ans.append(root.data)
            self.InOrder(root.right)
    def traversal(self, root):
        self.InOrder(root)
        return self.ans
```

```python
from queue import Queue #importing the Queue data structure 
class LevelOrderTraversal:
    def levelOrder(self, root):
        ans = []
        if root == None:
            return ans
        q = Queue(maxsize= 20)
        q.put(root) #enqueues the root to the queue
        ans.append([root.data])
        
        while q.size() > 0:
            l = q.size
            level = []
            for i in range(l):
                front = q.get()
                if front.left != None:
                    q.put(front.left)
                    level.append(front.left.data)
                if front.right != None:
                    q.put(front.right)
                    level.append(front.right.data)
                if len(level) > 0:
                    ans.append(level)
        return ans
    
```
